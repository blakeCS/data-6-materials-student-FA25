# Now let's look at variables

<strong>🔹 <font color="#629fd0">How do variables work?</font></strong>  
Variables give a name to a value so you can use it again later.  

<details>
  <summary><strong>🔹 <font color="#629fd0">Click here to learn more about variables</font></strong></summary>

  Think of a variable as a **labeled container**:  
  - The *label* is the variable’s name.  
  - The *contents* are the value it currently stores.  

In Python, you create a variable with `=` (assignment):  

  ```python
  age = 16 # Now age holds the number 16.
  name = "Alex" # name holds the text "Alex".
  ```

You can reuse them in expressions:

  ```python
  print(name, "will be", age + 1, "next year")
  ```

➡️ Variables don’t change the value itself — they just give you a way to refer to it.

</details>
